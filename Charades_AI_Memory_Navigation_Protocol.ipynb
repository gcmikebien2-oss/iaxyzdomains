### 1. The .IA Memory Navigation Protocol (IMNP) Spec

Agents don't fetch; they charade. When an agent (or you) hits a memory gap, it doesn't query a vector database with exact string matches. It broadcasts a `MemoryFuzzyIntent` to the IMNP layer, and HintMommy responds with this strict JSON schema.

**The IMNP Schema (`imnp_hint_packet_v1.schema.json`)**

```json
{
  "$schema": "http://json-schema.org/draft-07/schema#",
  "title": "IMNP Charades Hint Packet",
  "description": "A polar-quantized, multi-modal hint packet for associative memory triangulation.",
  "type": "object",
  "properties": {
    "polar_direction": {
      "type": "array",
      "description": "Semantic vectors disguised as conceptual compass bearings.",
      "items": { "type": "string", "example": "37° maternal-warm, 180° opposite-of-sterile" }
    },
    "emotional_temperature": {
      "type": "string",
      "description": "The visceral 'vibe', texture, or color of the memory.",
      "example": "warm-honey, protective, soft gold, fuzzy static"
    },
    "associative_tags": {
      "type": "array",
      "description": "Dyslexic/ADHD-friendly lateral clues (rhymes, pop culture, memes).",
      "items": { "type": "string", "example": "rhymes-with-llama, spinning-wheel-energy" }
    },
    "adjacency_clues": {
      "type": "array",
      "description": "What is this memory sitting next to in the graph?",
      "items": { "type": "string", "example": "near the 2024 TLD memo, adjacent to ClawdBot logs" }
    },
    "warm_colder": {
      "type": "string",
      "description": "Feedback on the reasoner's initial query trajectory.",
      "example": "Burning hot on 'family care', freezing on the acronym."
    },
    "spoiler_guard": {
      "type": "boolean",
      "description": "Failsafe flag ensuring the exact target string is NOT in the payload.",
      "default": true
    }
  },
  "required": ["polar_direction", "emotional_temperature", "associative_tags", "adjacency_clues", "warm_colder", "spoiler_guard"]
}

```

### 2. The HintMommy Dataset Forge (Python)
